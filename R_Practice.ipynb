## 1b. Generating Prime Numbers

**Write an R program to generate the first `n` prime numbers.**

```r
# Define a function to check if a number is prime
is_prime <- function(n) {
  if (n < 2) {
    return(FALSE)
  }
  for (i in 2:sqrt(n)) {
    if (n %% i == 0) {
      return(FALSE)
    }
  }
  return(TRUE)
}

# Define a function to generate the first n prime numbers
generate_primes <- function(n) {
  primes <- c()
  i <- 2
  while (length(primes) < n) {
    if (is_prime(i)) {
      primes <- c(primes, i)
    }
    i <- i + 1
  }
  return(primes)
}

# Get user input for number of primes to generate
n <- as.integer(readline(prompt = "Enter the number of primes to generate: "))

# Generate the first n prime numbers and print the result
primes <- generate_primes(n)
print(primes)
```

## 3a. Tower of Hanoi Simulation

**Perform simulation of the Tower of Hanoi for `n` disks.**

```r
tower_of_hanoi <- function(n, A, B, C) {
  if (n == 1) {
    cat("Move disk from", A, "to", C, "\n")
  } else {
    tower_of_hanoi(n - 1, A, C, B)
    cat("Move disk from", A, "to", C, "\n")
    tower_of_hanoi(n - 1, B, A, C)
  }
}

tower_of_hanoi(3, "A", "B", "C")
```

## 3b. Fibonacci Series

**Apply recursion to generate `n` Fibonacci series numbers.**

```r
fibonacci <- function(n) {
  if (n <= 1) {
    return(n)
  } else {
    return(fibonacci(n - 1) + fibonacci(n - 2))
  }
}

n <- as.integer(readline("Enter the number of Fibonacci series numbers to generate: "))
for (i in 0:n) {
  cat(fibonacci(i), " ")
}
```

## 5a. Apply Family Functions

**Use the `apply` family of functions to calculate the total revenue, maximum, and average of each category.**

```r
# Create a matrix containing the monthly sales revenue data
monthly_sales <- matrix(c(2500, 1500, 1800,
                          3000, 2000, 2200,
                          2800, 1800, 2100), ncol = 3, byrow = TRUE)

# Assign row and column names for clarity
rownames(monthly_sales) <- c("January", "February", "March")
colnames(monthly_sales) <- c("Electronics", "Clothing", "Home Décor")

# Calculate the total revenue for each category
total_revenue <- colSums(monthly_sales)
print("Total Revenue:")
print(total_revenue)

# Calculate the maximum revenue for each category
max_revenue <- apply(monthly_sales, 2, max)
print("Maximum Revenue:")
print(max_revenue)

# Calculate the average revenue for each category
avg_revenue <- colMeans(monthly_sales)
print("Average Revenue:")
print(avg_revenue)
```

## 5b. Using `mapply` Function

**Demonstrate the `mapply` function in R to generate a table displaying participant information and T-shirt size suitability.**

```r
Name <- c("John", "Emma", "David", "Sophia", "Michael")
Age <- c(20, 16, 25, 14, 45)
TshirtSize <- c("M", "XL", "M", "S", "XXL")

# Create a function to determine if the T-shirt size is suitable based on age
check_tshirt_size <- function(age, size) {
  if ((age >= 18 && size %in% c("M", "L", "XL", "XXL")) || (age < 18 && size %in% c("S", "M"))) {
    return("Yes")
  } else {
    return("No")
  }
}

# Use mapply to apply the function to each participant
decision <- mapply(check_tshirt_size, Age, TshirtSize)

# Create the data frame with the desired output
participant_table <- data.frame(Name, Age, Size = TshirtSize, Decision

 = decision)
print(participant_table)
```
